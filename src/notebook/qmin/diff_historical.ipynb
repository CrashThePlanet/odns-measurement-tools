```python
historical_path = "/mainData/projects/bachelors/odns-measurement-tools/src/data/raw/udp_dataframe_complete_historical.csv"
new_path = "/mainData/projects/bachelors/odns-measurement-tools/src/data/raw/odns_udp_resolver_2026-05-31.csv"
```

```python
import pandas as pd

hist = pd.read_csv(historical_path, sep=";")
new = pd.read_csv(new_path)
```

```python
hist = hist[hist["response_type"] == "Resolver"]

old_resolver_unaccounted = hist[~hist["ip_request"].isin(new["queried_ip"])]

old_resolver_unaccounted.drop("id", axis=1)

old_resolver_unaccounted.rename(columns={
    "ip_request": "queried_ip",
    "ip_response": "replying_ip",
    "a_record": "backend_resolver",
    "response_type": "resolver_type",
    "country_request": "queried_ip_country",
    "asn_request": "queried_ip_asn",
    "prefix_request": "queried_ip_prefix",
    "org_request": "queried_ip_org",
    "country_response": "replying_ip_country",
    "asn_response": "replying_ip_asn",
    "prefix_response": "replying_ip_prefix",
    "org_response": "replying_ip_org"
}, inplace=True)

old_resolver_unaccounted.head()

old_resolver_unaccounted.to_csv("/mainData/projects/bachelors/odns-measurement-tools/src/data/raw/old_unaccounted_resolver.csv", index=False)
```
